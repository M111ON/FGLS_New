```
def test_model_inference(llm_instance, prompt, model_name="Model"):
    print(f"--- Testing {model_name} ---")

    # Simple chat template application
    messages = [
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": prompt}
    ]

    try:
        # Generate response
        response = llm_instance.create_chat_completion(
            messages=messages,
            max_tokens=128,
            temperature=0.7
        )

        content = response['choices'][0]['message']['content']
        print(f"Response:\n{content}\n")
    except Exception as e:
        print(f"Error testing {model_name}: {e}")

# Example usage with all three models
test_prompt = "Explain what a black hole is in one sentence."

test_model_inference(llm_1, test_prompt, "SmolLM2-360M")
test_model_inference(llm_2, test_prompt, "Qwen2.5-1.5B")
test_model_inference(llm_3, test_prompt, "Qwen2.5-Coder-3B")
```

```
from mcp_client import MCPClient
import time

# Initialize client pointing to the local MCP server
mcp = MCPClient("http://0.0.0.0:8765")

print("Executing high-speed route resolution test...")
start_time = time.time()

count = 0
# Iterate through entity and locality space to verify O(1) lookup
for ent in range(0, 256, 10):
    for loc in range(0, 256, 30):
        # Resolve route for Zone 2 (S-Shape) as per the coord_real_registry
        r = mcp.resolve_route(2, ent, loc, 128)
        count += 1

end_time = time.time()
duration = end_time - start_time

print(f"Completed {count} resolutions in {duration:.4f} seconds.")
print(f"Average time per resolution: {(duration/count)*1000:.4f} ms")
if count > 0:
    print(f"Last resolution result: {r}")
```

```
import time
import numpy as np
from mcp_client import MCPClient

mcp = MCPClient("http://0.0.0.0:8765")

def run_benchmark(model_id, zones, iterations=100):
    print(f"Benchmarking Model: {model_id}...")
    latencies = []

    for _ in range(iterations):
        # Randomized entity and locality to prevent simple caching bias
        ent = np.random.randint(0, 255)
        loc = np.random.randint(0, 255)
        zone = np.random.choice(zones)

        start = time.perf_counter()
        _ = mcp.resolve_route(zone, ent, loc, 128)
        end = time.perf_counter()
        latencies.append((end - start) * 1000)

    avg_lat = np.mean(latencies)
    p95_lat = np.percentile(latencies, 95)
    print(f"  >> Avg Latency: {avg_lat:.4f} ms")
    print(f"  >> P95 Latency: {p95_lat:.4f} ms\n")
    return avg_lat, p95_lat

# SmolLM2 typically occupies Zones 6-8
smol_zones = [6, 7, 8]
# Qwen2.5 typically occupies Zones 0-3 (Coder) or 4, 5, 9-11 (Standard)
qwen_zones = [0, 1, 2, 4, 5]

results = {}
results['SmolLM2'] = run_benchmark('smollm2', smol_zones)
results['Qwen2.5-SF'] = run_benchmark('qwen25', qwen_zones)
```

### Stress Test: Throughput & Latency Distribution
เราจะทำการส่งคำถามแบบ Batch เพื่อวัดค่าความเร็วเฉลี่ย (Mean) และค่าความเสถียร (Variance) ของความเร็วในการสร้างข้อความ

```
import time

def benchmark_throughput(llm, model_name, prompt="Write a short paragraph about the future of AI."):
    print(f"--- Stress Testing: {model_name} ---")
    start_time = time.perf_counter()

    # Generate with fixed token limit to measure speed
    output = llm.create_chat_completion(
        messages=[{"role": "user", "content": prompt}],
        max_tokens=100,
        temperature=0.0 # Deterministic for benchmarking
    )

    end_time = time.perf_counter()
    duration = end_time - start_time

    tokens_generated = output['usage']['completion_tokens']
    tps = tokens_generated / duration

    print(f"Generated {tokens_generated} tokens in {duration:.2f} seconds")
    print(f"Throughput: {tps:.2f} tokens/sec")
    print(f"Full Latency: {duration*1000:.2f} ms\n")
    return tps

# Benchmark all loaded models
tps_results = {}
tps_results['SmolLM2'] = benchmark_throughput(llm_1, "SmolLM2-360M")
tps_results['Qwen2.5-1.5B'] = benchmark_throughput(llm_2, "Qwen2.5-1.5B")
tps_results['Qwen2.5-Coder-3B'] = benchmark_throughput(llm_3, "Qwen2.5-Coder-3B")
```

```
import matplotlib.pyplot as plt

# Visualize the efficiency gap
names = list(tps_results.keys())
values = list(tps_results.values())

plt.figure(figsize=(10, 5))
plt.bar(names, values, color=['#4CAF50', '#2196F3', '#FF9800'])
plt.title('Throughput Comparison (Tokens/Sec)')
plt.ylabel('Tokens Per Second')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```
from huggingface_hub import snapshot_download
import os

# Download Qwen2.5-1.5B-Instruct Safetensors
model_dir = snapshot_download(
    repo_id="Qwen/Qwen2.5-1.5B-Instruct",
    allow_patterns=["*.safetensors", "*.json"]
)

print(f"Model downloaded to: {model_dir}")
```

```
build_script_content = """
import os
import json
import torch
import argparse
from safetensors import safe_open

def build_store(model_path, model_name, out_dir):
    os.makedirs(out_dir, exist_ok=True)
    raw_dir = os.path.join(out_dir, f'{model_name}_tensors_raw')
    os.makedirs(raw_dir, exist_ok=True)

    gsidx = []
    offset = 0

    # List all safetensors files
    files = [f for f in os.listdir(model_path) if f.endswith('.safetensors')]

    for filename in sorted(files):
        file_path = os.path.join(model_path, filename)
        with safe_open(file_path, framework='pt', device='cpu') as f:
            for key in f.keys():
                tensor = f.get_tensor(key)

                # Handle BFloat16 conversion for NumPy compatibility
                is_bf16 = (tensor.dtype == torch.bfloat16)
                if is_bf16:
                    tensor = tensor.to(torch.float32)

                # Determine zone (Simplified logic for Qwen architecture)
                zone = 0
                if 'layers' in key:
                    try:
                        layer_num = int(key.split('layers.')[1].split('.')[0])
                        zone = layer_num % 12
                    except: pass

                # Save raw qdat (numpy format emulation)
                out_name = f'{key}.qdat'
                out_path = os.path.join(raw_dir, out_name)
                # Explicitly cast to numpy after potential dtype change
                tensor.detach().cpu().numpy().tofile(out_path)

                nbytes = tensor.element_size() * tensor.nelement()

                gsidx.append({
                    'name': key,
                    'shape': list(tensor.shape),
                    'dtype': 0 if not is_bf16 and tensor.dtype == torch.float32 else 1,
                    'offset': offset,
                    'nbytes': nbytes,
                    'path': os.path.relpath(out_path, out_dir),
                    'geo_key': hash(key) & 0xFFFFFFFFFFFFFFFF,
                    'zone': zone
                })
                offset += nbytes

    idx_path = os.path.join(out_dir, f'{model_name}_geom.gsidx')
    with open(idx_path, 'w') as f:
        json.dump(gsidx, f, indent=2)

    print(f'Done! {len(gsidx)} tensors processed.')
    print(f'Index: {idx_path}')

if __name__ == '__main__':
    parser = argparse.ArgumentParser()
    parser.add_argument('--model', required=True)
    parser.add_argument('--model-name', required=True)
    parser.add_argument('--out', default='build/')
    args = parser.parse_args()
    build_store(args.model, args.model_name, args.out)
"""

with open('build_from_safetensors.py', 'w') as f:
    f.write(build_script_content.strip())

print("Fixed build_from_safetensors.py created with explicit BFloat16 to Float32 conversion for NumPy.")
```
